```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
lent_sim = np.random.geometric(p = 1/7, size = 9*10**7)



plt.hist(lent_sim, bins = 40, density = True)
plt.xlim(right = 40)
plt.xlabel('days elapsed')
plt.ylabel('fraction of ppl')
plt.title('when are you gonna fold during lent?')
plt.show()
```
