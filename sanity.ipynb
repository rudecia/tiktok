probabilities:

p(bad day) = 0.1 (-2 sanity points)

p(normal day) = 0.8 (-1 sanity points)

p(good day) = 0.1 (+1 sanity point)


```python
import numpy as np
import matplotlib.pyplot as plt


N = 10**5

sanity_at_graduation = np.zeros(N)


for i in range(N):
  sanity_each_day = np.random.choice(a = (-2, -1, 1),  p = (0.1, 0.8, 0.1), size = 73)
  sanity_at_graduation[i] = 80 + sum(sanity_each_day)




plt.hist(sanity_at_graduation)
plt.title('Sanity Points Remaining at Graduation')
plt.show()
```
