A Simulation Study of Markov Chain Convergence

```python
import numpy as np



#transition matrix


N = 10**5

P =[[0.1, 0.4, 0.5],
    [0.3, 0.2, 0.5],
    [0.3, 0.3, 0.4]]


mu = np.array([0.3, 0.3, 0.4])


trials = 200



seq = np.zeros(N)

seq[0] = np.random.choice(a = [0, 1, 2], p = mu, size = 1)[0]


for i in range(1, N):
  prev = seq[i -1]
  pmf = P[int(prev)]
  seq[i] = np.random.choice(a = [0,1,2], p = pmf, size = 1)[0]




steady = np.array((np.mean(seq ==0), np.mean(seq == 1),  np.mean(seq ==2)))


print(steady)
```

```python

eigval , eigvect = np.linalg.eig(np.array(P).T)

eigvect[:, 0]/sum(eigvect[:, 0])

```
