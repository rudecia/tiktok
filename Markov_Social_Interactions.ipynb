# Using Markov Chains to Simulate the Flow of Gossip




```python
import numpy as np
import matplotlib.pyplot as plt


p_init = 1 #first person you told
p_dirt = 3 #person you have dirt on



P = np.array([
    [0.95, 0.01, 0.01, 0.03],
     [0.1, 0.8, 0.05, 0.05],
     [0.05, 0.05, 0.85, 0.05],
     [0.02, 0.02, 0.16, 0.8]])

N = 100

time_elapsed = np.zeros(N)
ppl = range(4)


for sim in range(N):
  next_person = p_init
  t = 0
  while next_person != p_dirt:
    dist = P[next_person, :]
    next_person = np.random.choice(a = ppl, p = dist, size = 1)[0]
    t +=1

  time_elapsed[sim] = t






plt.hist(time_elapsed)
plt.title('Time Until Someone Beats You Up For Gossiping')
plt.ylabel('Day Elapsed')
plt.show()
```
