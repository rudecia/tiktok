# Parameters

```python
p_the_one = 1/10**5 #probability of meeting the one in a given interaction
```

# Simulation Study

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 10**6


n_interactions = np.random.geometric(p_the_one, N) #how many interactions will it take to meet the one?

n_days = n_interactions/3 #rate of three interactions per day

n_years = n_days/365.25


plt.hist(n_years, bins = np.linspace(0, 200, 10), density = True)
plt.title('Years Necessary to Meet the One')
plt.xlabel('Years Elapsed')
plt.ylabel('Proportion of People')
plt.xlim(right = 200)
plt.show()


```

Probability of Meeting the One Before 40?

```python
before_40 = np.mean(n_years < 40)

print(f'The probability of meeting the one before age 40 is {np.round(before_40, 3)}')
```

More realistically, we meet more people in the prime of our lives than when we are older. A gamma distribution can capture this discrepancy

```python
#e(x) = shape * scale
#var(x) = shape * scale^2

interactions_gamma = np.random.gamma(scale = 11, shape = 2,size =  N)


plt.hist(interactions_gamma, bins = np.linspace(0, 120, 10) )
plt.show()



p_before_40_gamma = np.mean(interactions_gamma < 40)
print(p_before_40_gamma)
```
